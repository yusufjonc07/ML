# Dimensionality Reduction

- if there are millions of features in the dataset, it may take very long time to train a model
- that is why, reducing features (dimensions) is needed in an optimal way
- Example, in the MNIST dataset, pixels on the border are always white so we can drop that pixels (features), also neighboring pixels are highly correlated -> **merging** -> take the mean



- In high dimensional space, intances located mostly near dimension borders because they are sparce
- As a result, models like KNN perform poorly when number of features/dimensions are too much
- High dimensional datasets -> risk of being too sparce -> training instances are far from each. other
- In short, the more dimensions the training set has, the greater the risk of overfitting it.

## Main Approaches for Dimensionality reduction

### Projection

- a plane - a lower dimensional subspace of higher dimensional space 
- instances are projected perpendicular to a plane -> projection are new features but less than before
- for exmaple, 3D -> **2D. x1, x2, x3** become **z1, x2**
